```python
import numpy as np

# Student model sizes in MB
ms_dict = {
    'small': 2.25,
    'medium': 6.21,
    'large': 12.22
}
Mmin = min(ms_dict.values())
Mmax = max(ms_dict.values())

gamma = 0.25
epsilon = 1e-7

# Load confidence scores for each region
low_Cd = np.load('low_confidence_data.npy')
medium_Cd = np.load('medium_confidence_data.npy')
high_Cd = np.load('high_confidence_data.npy')

print("Confidence scores loaded.")

def calculate_alpha_d(Ms, Cd, gamma=1.0, epsilon=1e-7, Mmin=None, Mmax=None):
    # Normalize Ms
    norm_Ms = (Ms - Mmin) / (Mmax - Mmin)
    g_Ms = 0.2 + 0.6 * norm_Ms
    h_Cd = Cd + epsilon  # Cd is normalized
    alpha_d = g_Ms * (gamma * 1 / h_Cd)
    # Optionally, clip for display
    alpha_d = np.clip(alpha_d, 0, 1)
    return alpha_d

alpha_results = {}

for model_size, Ms_val in ms_dict.items():
    alpha_results[model_size] = {}
    alpha_low = calculate_alpha_d(Ms_val, low_Cd, gamma, epsilon, Mmin, Mmax)
    alpha_medium = calculate_alpha_d(Ms_val, medium_Cd, gamma, epsilon, Mmin, Mmax)
    alpha_high = calculate_alpha_d(Ms_val, high_Cd, gamma, epsilon, Mmin, Mmax)
    alpha_results[model_size]['low'] = np.median(alpha_low)
    alpha_results[model_size]['medium'] = np.median(alpha_medium)
    alpha_results[model_size]['high'] = np.median(alpha_high)

for model_size in alpha_results:
    print(f"\nAlpha values for {model_size} student model:")
    print(f"  Low confidence region alpha: {alpha_results[model_size]['low']:.4f}")
    print(f"  Medium confidence region alpha: {alpha_results[model_size]['medium']:.4f}")
    print(f"  High confidence region alpha: {alpha_results[model_size]['high']:.4f}")
```

```python
def map_alpha(value):
    if value <= 0.3:
        return 0.3
    elif value >= 0.7:
        return 0.7
    else:
        return 0.5

for model_size in alpha_results:
    print(f"\nAlpha values for {model_size} student model:")
    print(f"  Low confidence region alpha: {map_alpha(alpha_results[model_size]['low']):.1f}")
    print(f"  Medium confidence region alpha: {map_alpha(alpha_results[model_size]['medium']):.1f}")
    print(f"  High confidence region alpha: {map_alpha(alpha_results[model_size]['high']):.1f}")
```
